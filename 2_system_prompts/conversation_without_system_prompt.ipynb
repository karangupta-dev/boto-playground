# Conversation without a system prompt

This example sends a multi-turn conversation to Bedrock without a system prompt. The `chat` helper's default empty `system_prompt` is used.

```python
import sys
from pathlib import Path

# Add the repository root when the notebook runs from the subdirectory
repo_root = Path.cwd().parent
sys.path.insert(0, str(repo_root))

from util import add_user_message, add_assistant_message, chat
```

```python
messages = []
add_user_message(messages, "What's 1 + 1?")

# No system_prompt argument is passed.
answer = chat(messages)
print(answer)
add_assistant_message(messages, answer)
```

```python
add_user_message(messages, "And 3 more added to that?")
answer = chat(messages)
print(answer)
```
