# Conversation with a system prompt

Compare an AWS question with an unrelated question. The system prompt instructs the assistant to answer only AWS cloud hosting questions.

```python
import sys
from pathlib import Path

# Add the repository root when the notebook runs from the subdirectory
repo_root = Path.cwd().parent
sys.path.insert(0, str(repo_root))

from util import add_user_message, chat

system_prompt = """
You are an AWS cloud support specialist. Your job is to answer user queries related 
to cloud hosting services on AWS.
"""
```

```python
messages = []
add_user_message(messages, "How can I host a static website using Amazon S3?")
answer = chat(messages, system_prompt)
print(answer)
```
