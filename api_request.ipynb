```python
import boto3

client = boto3.client("bedrock-runtime", region_name="us-east-2")

model_id = "us.anthropic.claude-sonnet-4-6"

user_message = {
    "role": "user",
    "content": [
        {"text": "What's 1+1?"}
    ]
}

response = client.converse(
    modelId=model_id,
    messages=[user_message]
)

```

```python
response["output"]["message"]["content"][0]["text"]
```
